```python
#install packages for this notebook
#%pip install pandas numpy scipy openpyxl

#First we will optimize a and m.
#Import the 3 packages you'll need
import pandas as pd
import numpy as np
from scipy.optimize import least_squares

#Load in the data as a pandas dataframe
por_F_in = pd.read_excel('por_vs_F.xlsx',0)
#Convert the columns of the dataframe to arrays for calculation
por_in = np.array(por_F_in['por_in'])
F_in = np.array(por_F_in['F_in'])

#Define the objective function that we want to minimize.
#In this case, we will calculate F using porosity, a, and m,
#and then compute the difference between that value and the true value of F.
#The variable archie_params is a two-element array where the first element
#is a and the second element is m.
def f_calc(archie_params):
    return np.abs(archie_params[0]/(por_in**archie_params[1]) - F_in)

#Define the initial guesses for a and m.
x0=[1,2]

#Here, we find the values of a and m that minimize our defined objective function using least squares
#optimization. We need to specify the initial guesses for a and m.
res1=least_squares(f_calc,x0)

#The results are stored in an object we have called res1. It has many attributes relating
#to the optimization, which you can view with the command print(res1). Here we are only
#going to print the optimized values of a and m, which are contained in a two-element array.
print(res1.x)
```

```python
#Next we will optimize n.
#Load in the values of water saturation and resistivity index to a pandas dataframe.
sw_I_R_in = pd.read_excel('Sw_vs_I_R.xlsx',0)
#Convert the columns of the dataframe to arrays for calculation.
sw_in = np.array(sw_I_R_in['Sw'])
I_R_in = np.array(sw_I_R_in['I_R'])

#Define the objective function you want to minimize.
def I_R_calc(n):
    return np.abs(1/ (sw_in**n) - I_R_in)

x0 = [2]

res2=least_squares(I_R_calc, x0)

#This command will print the optimized value of n, which is stored in an object called res2.
print(res2.x)
```
